First CNN Project: Project Fashion Classification

Improt libreries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import keras
```

Load Data

```python
(X_train,y_train),(X_test,y_test)=keras.datasets.fashion_mnist.load_data()
```

```python
X_train.shape,y_train.shape,X_test.shape,y_test.shape
```

```python
class_names = ['T-shirt/top', 'Trouser', 'Pullover', 'Dress', 'Coat', 'Sandal', 'Shirt', 'Sneaker', 'Bag', 'Ankle boot']
print(f"The label for the first training image (y_train[0]) is: {class_names[y_train[0]]}")
```

The Fashion MNIST dataset consists of 10 classes, each represented by a numerical label from 0 to 9. The code above maps the numerical label to its corresponding class name, making the output more descriptive.

change dimention

```python
X_train.ndim
```

```python
X_train.shape
```

```python
X_train=np.expand_dims(X_train, -1)
X_test=np.expand_dims(X_test, -1)
```

```python
X_train.ndim
```

```python
X_train.shape
```

Feature Scaling

```python
X_train=X_train/255.0
X_test=X_test/255.0
```

```python
X_train[0]
```

Split Datasets

```python
from sklearn.model_selection import train_test_split
X_train, X_val, y_train, y_val = train_test_split(X_train, y_train, test_size=0.2,random_state=2020)
```

```python
X_train.shape, y_train.shape
```

```python
X_val.shape, y_val.shape
```

Convolution netural network-model Building

```python
model = keras.models.Sequential([
                         keras.layers.Conv2D(filters=32, kernel_size=(3,3),strides=(1, 1),padding='valid', activation='relu', input_shape=(28,28,1)),
                         keras.layers.MaxPooling2D(pool_size=(2,2)),
                         keras.layers.Flatten(),
                         keras.layers.Dense(units=128, activation='relu'),
                         keras.layers.Dense(units=10, activation='softmax')

])
```

```python
model.summary()
```

```python
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
```

```python
model.fit(X_train, y_train, epochs=10, batch_size=512, verbose=1, validation_data=(X_val, y_val))
```

Test the model

```python
model.predict(np.expand_dims(X_test[0], axis=0)).round(2)
```

```python
np.argmax(model.predict(np.expand_dims(X_test[0], axis=0))).round(2)
```

```python
y_test[0]
```

```python
y_pred = model.predict(X_test).round(2)
y_pred
```

```python
model.evaluate(X_test, y_test)
```

Visualize Output

```python
plt. figure(figsize=(16,30))

j=1
for i in np.random.randint(0, 1000, 50): # Changed 60 to 50
 plt. subplot(10,5,j); j+=1
 plt.imshow(X_test[i].reshape(28,28), cmap="Greys")
 plt.axis('off') # off the axis
 plt.title('Actual = {} / {} \nPredicted = {} / {}'.format(class_names[y_test[i]], y_test[i], class_names[np.argmax(y_pred[i])], np.argmax(y_pred[i])))
```

confusion matrix

```python
from sklearn.metrics import confusion_matrix

plt.figure(figsize=(16, 9))
y_pred_labels = [ np.argmax(label) for label in y_pred ]
cm = confusion_matrix(y_test, y_pred_labels)

#show cm
sns.heatmap(cm, annot=True, fmt='d',xticklabels=class_names, yticklabels=class_names)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, y_pred_labels, target_names=class_names))
```
